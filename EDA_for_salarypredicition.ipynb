### Data Cleaning

```python
# drop column unnamed
df.drop(labels='Unnamed: 0', axis='columns',inplace=True)
df.head(1)
```

```python
# Some rows having inconsistent value as -1 in salary estimate columns.
print("Before : ",df.shape)
df=df[df['Salary Estimate']!="-1"]
print("After : ",df.shape)
```

```python
# 72𝐾−133K (Glassdoor est.) Need to remove (Glassdoor est.) textual data from salary estimate column.

salary= df['Salary Estimate'].apply(lambda x: x.split("(")[0])

salary
```

```python
# Remove specialcharacters 'K' '$' from salary estimate.
salary= salary.apply(lambda x: x.replace("$","").replace("K",""))
salary
```

### Remove Dupliactes

```python
# our data set doesnot contain duplicates but below is the code to remove duplicates
df.drop_duplicates(inplace=True) 
```

### Remove data inconsistencies

```python
# Finding any inconsistency in the salary
print("Length of salary: ", len(salary.unique()))
salary.unique()[380:]
```

```python
# Creating Column for 'per hour salary'
df['salary_per_hour'] = salary.apply( lambda x: 1 if "Per Hour" in x.lower() else 0 )
```

```python
df['salary_per_hour'].value_counts()
```

```python
# Creating Column for "Employer Provided Salary"
df['emp_provided_salary']= salary.apply(lambda x: 1 if "employer provided salary" in x.lower() else 0)
df["emp_provided_salary"].value_counts()
```

```python
# Removing 'per hour' and 'Employer provided salary'from ' salary estimate' 
salary = salary.apply(lambda x : x.lower().replace("per hour","").replace("employer provided salary:","").replace("",""))
salary.unique()[380:]
```

```python
# Creating column for min_salary
df["min_salary"]= salary.apply(lambda x: int(x.split("-")[0]))
df["min_salary"].tail()
```

```python
# Creating column for max_salary
df["max_salary"]= salary.apply(lambda x: int(x.split("-")[1]))
df["max_salary"].tail()
```

```python
#Finding avg_salary
df["avg_salary"]= (df['min_salary'] +df['max_salary'])/2
df["avg_salary"].tail()
```

```python
# Removing number from "Company name " column
df["Company Name"] = df["Company Name"].apply(lambda x : x.split("\n") [0])
df["Company Name"].head()
```

```python
# Creating a column for "Job state"
df["job_state"] = df["Location"].apply(lambda x : x.split(',')[1])
df["job_state"].head()
```

```python
df['job_state'].unique()
```

```python
# Fixing Los Angeles to CA
df['job_state']=df ['job_state'].apply(lambda x: x.strip() if x.strip().lower() !=' los angeles' else 'CA')
df['job_state'].value_counts()[:5]
```

```python
# Caluculating age of the companies
df["company_age"]= df['Founded'].apply(lambda x: x if x<1 else 2023-x)
df['company_age'].head()
```

```python
# Clearing 'type of ownership' column
df['Type of ownership'].value_counts()
```

```python
#need to simplify type
def ownership_simplifier(text):
    if 'Private' in text.lower():
        return 'Private'
    elif 'Public' in text.lower():
        return 'Public'
    elif ('-1'in text.lower()) or ('Unknown' in text.lower()):
        return "Other Organization"
    else: 
        return text
df["Type of ownership"]=df["Type of ownership"].apply(ownership_simplifier)
df['Type of ownership'].value_counts()
```

```python
# Clearing 'Revenue' column
df['Revenue'].value_counts()
```

```python
def revenue_simplifier(text):
    if '-1' in text.lower():
        return 'Non-applicable'
    else: 
        return text
df["Revenue"]=df["Revenue"].apply(revenue_simplifier)
df['Revenue'].value_counts()
```

```python
def industry_simplifier(text):
    if '-1' in text.lower():
        return 'Non-applicable'
    else: 
        return text
df["Industry"]=df["Industry"].apply(revenue_simplifier)
df['Industry'].value_counts()
```

```python
def sector_simplifier(text):
    if '-1' in text.lower():
        return 'Non-applicable'
    else: 
        return text
df["Sector"]=df["Sector"].apply(revenue_simplifier)
df['Sector'].value_counts()
```
